## Scoping Paragraph
This projects scope is to understand whether an ETF trading strategy that has performed well on historical data will continue to be effective going forward,the core question a portfolio manager faces before committing further capital to it. Because market conditions and patterns can shift over time, a strategy built on past price behavior carries the risk that the conditions it relied on may no longer hold, meaning strong historical performance does not guarantee future results. Understanding this risk is critical before continuing capital allocation to the strategy, since a rule that looked profitable in a backtest could underperform once market conditions change.


# Stakeholder Brief — ETF Strategy Viability Assessment
**Audience:** Portfolio Manager | **Cadence:** Monthly, or before rebalancing decisions
**Decision Supported:** Continue / adjust / discontinue capital allocation to the strategy

## Context
The strategy has performed well historically, but continued allocation requires confidence it will keep performing under current and future market conditions.

## What You'll Receive
- Forecast and risk bands for expected performance over the next year
- Scenario analysis under different market conditions
- Plain-language summary of assumptions and risks

## Assumptions & Constraints
- Historical patterns are assumed to persist (stationarity risk if they don't)
- ETF assumed liquid enough to trade without moving its own price
- Backtested transaction costs assumed representative of live costs
- Analysis bounded to the historical data window tested
